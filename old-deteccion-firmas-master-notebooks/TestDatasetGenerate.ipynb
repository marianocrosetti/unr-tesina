### Extrayendo firmas

### Enderazamos, recortamos y escalamos 

```python
pth = './raw_imgs/test/'
dst_pth = './telegramas-crop-imgs'
fns = os.listdir(pth)
for fn in fns:
    img = Image.open(os.path.join(pth,fn))
    pimg = preprocess_pipeline(img)
    pimg.save(os.path.join(dst_pth, fn))
```

### Binarizamos

```python
fns=get_image_files('./dataset-test/cls-margin-no-preproc')
for fn in fns:
    cls = fn.parent.name
    result_fn = Path(str(fn.parent.parent).replace('no-preproc','preproc')) / cls / fn.name
    img = Image.open(fn)
    pimg = preproc(img)
    pimg.save(create_if_not_exists(result_fn))
```
